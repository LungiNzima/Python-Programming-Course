```python
import numpy as np
```

```python
arr_1d = np.array([1,2,3,4,5])
print(arr_1d)
```

```python
arr_2d = np.array([[1,2,3],[4,5,6]])
print(arr_2d)
```

```python
element = arr_1d[2]
print(element)
```

```python
row = arr_2d[1]
print(row)
```

```python
specific_element = arr_2d[0,2]
print(specific_element)
```

```python
subarray = arr_1d[1:4]
print(subarray)
```

```python
reshaped_arr = arr_1d.reshape((5,1))
print(reshaped_arr)
```

```python
flattened_arr = arr_2d.flatten()
print(flattened_arr)
```

```python
concatenated_arr = np.concatenate([arr_1d,arr_1d])
print(concatenated_arr)
```

```python
concatenated_arr_2d = np.concatenate([arr_2d,arr_2d], axis = 1)
print(concatenated_arr_2d)
```

```python
split_arr = np.split(arr_1d, 5)
print(split_arr)
```

```python
split_arr_2d = np.split(arr_2d,2,axis = 0)
print(split_arr_2d)
```

```python
transposed_arr = arr_2d.T
print(transposed_arr)
```

```python
addition = arr_1d + 2
print(addition)
```

```python
multiplication = arr_2d * 2
print(multiplication)
```

```python
#sort 1d array in ascending order
sorted_arr = np.sort(arr_1d)
print(sorted_arr)
```

```python
sorted_arr_2d = np.sort(arr_2d, axis=0)
print(sorted_arr_2d)
```

```python
sorted_arr_2d_column = arr_2d[arr_2d[:,1].argsort()]
print(sorted_arr_2d_column)
```

```python
mask = arr_2d > 3
print(mask)
```

```python
filtered_arr = arr_2d[mask]
print(filtered_arr)
```

```python
arr_2d[mask] = 0
print(arr_2d)
```

```python
hstacked_arr = np.hstack([arr_2d, arr_2d])
print(hstacked_arr)
```

```python
vstacked_arr = np.vstack([arr_2d,arr_2d])
print(vstacked_arr)
```
