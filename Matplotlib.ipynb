```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,10,100)
y = np.sin(x) + np.random.normal(0,0.2,100)
```

```python
plt.plot(x,y)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
plt.scatter(x,y)
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
categories = ["A","B","C","D","E"]
values = [15,7,10,12,8]
plt.bar(categories, values)
plt.title("Bar Plot")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
plt.hist(y, bins = 20)
plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

```python
labels = ["Category A","Category B","Category C"]
sizes = [40,30,30]
explode = [0.1,0,0]
```

```python
plt.pie(sizes,explode = explode, labels = labels)
plt.title("Pie Chart")
plt.show()
```

```python
plt.fill_between(x,y,alpha = 0.5)
plt.title("Area Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```
